```python
#fission cross section
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from mpl_toolkits.axes_grid1.inset_locator import inset_axes, mark_inset

def load(filename):
    df = pd.read_csv(filename, header=None, names=["Energy", "CrossSection"])
    df["Energy"] = pd.to_numeric(df["Energy"], errors="coerce")
    df["CrossSection"] = pd.to_numeric(df["CrossSection"], errors="coerce")
    df = df.dropna()
    df["Energy"] = df["Energy"] / 1e6
    df = df.sort_values("Energy")
    return df

fission_isotopes = [
    "U233", "U235", "U238", "Pu239", "Pu240", "Pu241",
    "Np237", "Th227", "Th229", "Am241", "Am242m", "Am243"
]

colors = cm.tab20(range(len(fission_isotopes)))

# ---- Select isotopes that rise at high energy (data-driven, not guessed) ----
threshold_isotopes = []
for iso in fission_isotopes:
    df = load(f"{iso}.csv")
    low_e_val = df[df["Energy"] < 1e-6]["CrossSection"].mean()
    high_e_val = df[(df["Energy"] >= 1) & (df["Energy"] <= 20)]["CrossSection"].max()
    if high_e_val > low_e_val:
        threshold_isotopes.append(iso)

# ---- Main figure (now wider) ----
fig, ax = plt.subplots(figsize=(14, 7))
for i, iso in enumerate(fission_isotopes):
    df = load(f"{iso}.csv")
    ax.plot(df["Energy"], df["CrossSection"], label=iso, color=colors[i])

ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Incident Energy (MeV)")
ax.set_ylabel("Fission Cross Section (b)")
ax.set_title("")


ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.12), ncol=6, frameon=False, fontsize=9)
axins = inset_axes(ax, width="22%", height="29%", loc="lower right",
                    bbox_to_anchor=(-0.009, 0.035, 1, 1), bbox_transform=ax.transAxes)
for iso in threshold_isotopes:
    i = fission_isotopes.index(iso)
    df = load(f"{iso}.csv")
    axins.plot(df["Energy"], df["CrossSection"], color=colors[i], label=iso)

axins.set_xlim(1, 20)
axins.set_ylim(0.3, 5)          # <-- tightened range reveals separation between lines
axins.set_xscale("log")
axins.set_yscale("log")
axins.tick_params(labelsize=7)
axins.legend(loc="upper center", ncol=2, fontsize=9, frameon=False, framealpha=1, edgecolor="gray")
mark_inset(ax, axins, loc1=2, loc2=2, fc="none", ec="gray", lw=0.8)
plt.savefig("Fission Cross Section")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from mpl_toolkits.axes_grid1.inset_locator import inset_axes, mark_inset

def load(filename):
    df = pd.read_csv(filename, header=None, names=["Energy", "CrossSection"])
    df["Energy"] = pd.to_numeric(df["Energy"], errors="coerce")
    df["CrossSection"] = pd.to_numeric(df["CrossSection"], errors="coerce")
    df = df.dropna()
    df["Energy"] = df["Energy"] / 1e6
    df = df.sort_values("Energy")
    return df

fission_isotopes = [
    "U233", "U235", "U238", "Pu239", "Pu240", "Pu241",
    "Np237", "Th227", "Th229", "Am241", "Am242m", "Am243"
]
colors = cm.tab20(range(len(fission_isotopes)))

elements = ["Be", "C", "H1", "H2", "O", "Cl35", "Cl37", "Na", "F"]
element_colors = cm.tab10(range(len(elements)))

threshold_isotopes = []
for iso in fission_isotopes:
    df = load(f"{iso}.csv")
    low_e_val = df[df["Energy"] < 1e-6]["CrossSection"].mean()
    high_e_val = df[(df["Energy"] >= 1) & (df["Energy"] <= 20)]["CrossSection"].max()
    if high_e_val > low_e_val:
        threshold_isotopes.append(iso)

# ---- Wider panel (a) via GridSpec ----
fig = plt.figure(figsize=(24, 7))
gs = fig.add_gridspec(1, 3, width_ratios=[1.6, 1, 1])
ax_fission = fig.add_subplot(gs[0])
ax_elastic = fig.add_subplot(gs[1])
ax_capture = fig.add_subplot(gs[2])

label_kwargs = dict(fontsize=15, fontweight="bold", va="bottom", ha="left")
legend_kwargs = dict(frameon=False, fontsize=7, prop={"weight": "bold"})

# ---- (a) Fission ----
for i, iso in enumerate(fission_isotopes):
    df = load(f"{iso}.csv")
    ax_fission.plot(df["Energy"], df["CrossSection"], label=iso, color=colors[i])
ax_fission.set_xscale("log")
ax_fission.set_yscale("log")
ax_fission.set_xlabel("Incident Energy (MeV)")
ax_fission.set_ylabel("Fission Cross Section (b)")
ax_fission.legend(loc="upper center", bbox_to_anchor=(0.5, -0.15), ncol=6, **legend_kwargs)
ax_fission.text(-0.08, 1.05, "(a)", transform=ax_fission.transAxes, **label_kwargs)

axins = inset_axes(ax_fission, width="30%", height="29%", loc="lower right",
                    bbox_to_anchor=(-0.01, 0.03, 1, 1), bbox_transform=ax_fission.transAxes)
for iso in threshold_isotopes:
    i = fission_isotopes.index(iso)
    df = load(f"{iso}.csv")
    axins.plot(df["Energy"], df["CrossSection"], color=colors[i], label=iso)
axins.set_xlim(1, 20)
axins.set_ylim(0.3, 5)
axins.set_xscale("log")
axins.set_yscale("log")
axins.tick_params(labelsize=6)
axins.legend(loc="upper center", ncol=2, fontsize=6, frameon=False, prop={"weight": "bold"})
mark_inset(ax_fission, axins, loc1=2, loc2=2, fc="none", ec="gray", lw=0.8)

# ---- (b) Elastic ----
for i, el in enumerate(elements):
    elastic = load(f"{el}_elastic.csv")
    ax_elastic.plot(elastic["Energy"], elastic["CrossSection"], color=element_colors[i], label=el)
ax_elastic.set_xscale("log")
ax_elastic.set_yscale("log")
ax_elastic.set_xlabel("Incident Energy (MeV)")
ax_elastic.set_ylabel("Elastic Cross Section (b)")
ax_elastic.legend(loc="upper center", bbox_to_anchor=(0.5, -0.15), ncol=5, **legend_kwargs)
ax_elastic.text(-0.08, 1.05, "(b)", transform=ax_elastic.transAxes, **label_kwargs)

# ---- (c) Capture ----
for i, el in enumerate(elements):
    capture = load(f"{el}_capture.csv")
    ax_capture.plot(capture["Energy"], capture["CrossSection"], color=element_colors[i], label=el)
ax_capture.set_xscale("log")
ax_capture.set_yscale("log")
ax_capture.set_xlabel("Incident Energy (MeV)")
ax_capture.set_ylabel("Capture Cross Section (b)")
ax_capture.legend(loc="upper center", bbox_to_anchor=(0.5, -0.15), ncol=5, **legend_kwargs)
ax_capture.text(-0.08, 1.05, "(c)", transform=ax_capture.transAxes, **label_kwargs)

plt.tight_layout()
plt.savefig("cross_sections_combined.png", bbox_inches="tight", dpi=300, transparent=True)
plt.show()
```
