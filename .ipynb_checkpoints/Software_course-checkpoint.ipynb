```python

import re
import sys
from pathlib import Path

import numpy as np
import pandas as pd

# ------------------------------ settings ------------------------------
INPUT_ENERGY_UNIT = "eV"          # unit of column 1 in your files ("eV", "keV" or "MeV")
TRAIN_MIN_ENERGY_MEV = 1e-3       # 1 keV: above this the curves are smooth (below: resonance spikes)
RECURSIVE = False                 # True = also search sub-folders
DROP_DUPLICATES = False           # False = keep every row exactly as in your files (repeated energies are real)

OUT_LONG, OUT_TRAIN = "all_data.csv", "train_data.csv"
OUTPUT_NAMES = {OUT_LONG, OUT_TRAIN, "dataset.csv", "dataset_long.csv", "dataset_train.csv", "dataset_wide.csv"}
BASE_COLUMNS = ["nuclide", "Z", "A", "N", "reaction", "energy_eV", "xs_barn"]

UNIT_TO_EV = {"ev": 1.0, "kev": 1e3, "mev": 1e6}

# ------------------------------ elements ------------------------------
# Z = 1..103 (stops before Mt so a file part like "mt18" is never read as an element)
SYMBOLS = ("H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni "
           "Cu Zn Ga Ge As Se Br Kr Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe "
           "Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu Hf Ta W Re Os Ir Pt Au Hg "
           "Tl Pb Bi Po At Rn Fr Ra Ac Th Pa U Np Pu Am Cm Bk Cf Es Fm Md No Lr").split()
Z_OF = {s: i + 1 for i, s in enumerate(SYMBOLS)}

FISSION_WORDS = {"fission", "fiss", "fis", "nf", "mt18", "mt018"}
CAPTURE_WORDS = {"capture", "capt", "cap", "ng", "ngamma", "mt102", "mt0102", "gamma", "radcapture"}
REACTION_ORDER = {"fission": 0, "capture": 1}


# ------------------------- file-name parsing -------------------------
def plausible(z, a):
    return z <= a <= 3.5 * z + 10


def find_nuclide(stem):
    """'U233_fission' -> ('U', 233, ''); 'Th_232_capture' -> ('Th', 232, ''); 'Am242m_cap' -> ('Am', 242, 'm')."""
    tokens = [t for t in re.split(r"[^A-Za-z0-9]+", stem) if t]
    cands = []
    for t in tokens:                                            # U233, u233, Am242m, 233U
        m = re.fullmatch(r"([A-Za-z]{1,2})(\d{1,3})(m\d?)?", t)
        if m:
            cands.append((m.group(1), m.group(2), m.group(3) or ""))
        m = re.fullmatch(r"(\d{1,3})([A-Za-z]{1,2})(m\d?)?", t)
        if m:
            cands.append((m.group(2), m.group(1), m.group(3) or ""))
    for t1, t2 in zip(tokens, tokens[1:]):                      # Th_232, Pu-239
        if re.fullmatch(r"[A-Za-z]{1,2}", t1):
            m = re.fullmatch(r"(\d{1,3})(m\d?)?", t2)
            if m:
                cands.append((t1, m.group(1), m.group(2) or ""))
    for sym, a, meta in cands:
        sym = sym.capitalize()
        if sym in Z_OF and plausible(Z_OF[sym], int(a)):
            return sym, int(a), meta.lower()
    return None


def find_reaction(stem):
    tokens = {t.lower() for t in re.split(r"[^A-Za-z0-9]+", stem) if t}
    if tokens & FISSION_WORDS:
        return "fission"
    if tokens & CAPTURE_WORDS:
        return "capture"
    low = stem.lower()
    if "fiss" in low:
        return "fission"
    if "captur" in low:
        return "capture"
    return None


# --------------------------- file reading ---------------------------
def read_two_columns(path):
    """Read a header-less 2-column file (comma, semicolon, tab or spaces). Non-numeric lines are dropped."""
    df = pd.read_csv(path, sep=r"[,;\s]+", engine="python", header=None, skipinitialspace=True)
    df = df.dropna(axis=1, how="all")
    if df.shape[1] < 2:
        raise ValueError("file has fewer than 2 columns")
    df = df.iloc[:, :2]
    df.columns = ["energy", "xs"]
    df["energy"] = pd.to_numeric(df["energy"], errors="coerce")
    df["xs"] = pd.to_numeric(df["xs"], errors="coerce")
    n_before = len(df)
    df = df.dropna()
    return df, n_before - len(df)


def process_file(path):
    stem = path.stem
    nuc = find_nuclide(stem)
    if nuc is None:
        return None, "no nuclide found in the file name"
    reaction = find_reaction(stem)
    if reaction is None:
        return None, "no reaction (fission / capture) found in the file name"
    sym, a, meta = nuc
    z = Z_OF[sym]
    name = f"{sym}{a}{meta}"
    try:
        df, n_dropped = read_two_columns(path)
    except Exception as exc:
        return None, f"could not read: {exc}"
    if df.empty:
        return None, "no numeric rows"
    e_ev = df["energy"].to_numpy(float) * UNIT_TO_EV[INPUT_ENERGY_UNIT.lower()]
    out = pd.DataFrame({
        "nuclide": name, "Z": z, "A": a, "N": a - z, "reaction": reaction,
        "energy_eV": e_ev, "energy_MeV": e_ev / 1e6, "xs_barn": df["xs"].to_numpy(float)})
    info = dict(file=path.name, nuclide=name, reaction=reaction, rows=len(out),
                emin=e_ev.min(), emax=e_ev.max(), dropped=n_dropped)
    return (out, info), None


# ------------------------------- main -------------------------------
def main():
    try:
        folder = Path(__file__).resolve().parent      # folder where this script is saved
    except NameError:                                 # e.g. Jupyter / interactive: use current folder
        folder = Path.cwd()
    pattern = "**/*.csv" if RECURSIVE else "*.csv"
    files = sorted(p for p in folder.glob(pattern) if p.name not in OUTPUT_NAMES)
    print(f"Found {len(files)} CSV files in {folder.resolve()}\n")

    frames, infos, skipped, sources, expected = [], [], [], {}, {}
    for p in files:
        res, err = process_file(p)
        if err:
            skipped.append((p.name, err))
            continue
        df, info = res
        frames.append(df)
        infos.append(info)
        expected.setdefault((info['nuclide'], info['reaction']), []).append(df['energy_eV'].to_numpy(float))
        sources.setdefault((info["nuclide"], info["reaction"]), []).append(p.name)
        flag = f"  (dropped {info['dropped']} non-numeric lines)" if info["dropped"] else ""
        print(f"OK  {p.name:32s} -> {info['nuclide']:8s} {info['reaction']:8s} "
              f"{info['rows']:7d} rows   E = {info['emin']:.3g} .. {info['emax']:.3g} eV{flag}")

    if not frames:
        print("\nNothing could be read.")
        print_skipped(skipped)
        return

    data = pd.concat(frames, ignore_index=True)
    n_all = len(data)
    if DROP_DUPLICATES:
        data = data.drop_duplicates()
    n_dup = n_all - len(data)
    # group the files (fission first, then capture; by Z, A) but NEVER touch the row order inside a file:
    # the energies stay exactly in the sequence they have in your CSV.
    data["_r"] = data.reaction.map(REACTION_ORDER)
    data = (data.sort_values(["_r", "Z", "A", "nuclide"], kind="stable")
                .drop(columns="_r").reset_index(drop=True))
    data[BASE_COLUMNS].to_csv(folder / OUT_LONG, index=False)

    # ---- verify: energies in the output are exactly the energies of your files, in the same order
    if not DROP_DUPLICATES:
        back = pd.read_csv(folder / OUT_LONG)
        ok = True
        for (n, r), arrays in expected.items():
            got = back.loc[(back.nuclide == n) & (back.reaction == r), "energy_eV"].to_numpy(float)
            want = np.concatenate(arrays)
            if len(got) != len(want) or not np.array_equal(got, want):
                ok = False
                print(f"VERIFY FAILED for {n} {r}")
        print("VERIFIED: energy values and their order are identical to your CSV files." if ok else
              "WARNING: output differs from input, see messages above.")

    # ---- training file
    train = data[(data.xs_barn > 0) & (data.energy_MeV >= TRAIN_MIN_ENERGY_MEV)].copy()
    train["log10_E_MeV"] = np.log10(train.energy_MeV)
    train["log10_xs_barn"] = np.log10(train.xs_barn)
    train[BASE_COLUMNS + ["log10_E_MeV", "log10_xs_barn"]].to_csv(folder / OUT_TRAIN, index=False)

    # ---- report
    def nuc_key(n):
        return (Z_OF[re.match(r"[A-Z][a-z]?", n).group(0)], int(re.search(r"\d+", n).group(0)), n)

    print("\n================== REPORT ==================")
    print(f"Files read             : {len(infos)}   skipped: {len(skipped)}")
    print(f"Rows in {OUT_LONG:14s}: {len(data)}   (duplicate rows removed: {n_dup})")
    print(f"Rows in {OUT_TRAIN:14s}: {len(train)}   (energy >= {TRAIN_MIN_ENERGY_MEV} MeV and xs > 0)")
    print(f"Rows with xs <= 0 kept in long file: {int((data.xs_barn <= 0).sum())}")
    print("\nNuclides found:", ", ".join(sorted(data.nuclide.unique(), key=nuc_key)))

    count_all = data.groupby(["nuclide", "reaction"]).size().unstack(fill_value=0)
    count_trn = train.groupby(["nuclide", "reaction"]).size().unstack(fill_value=0)
    count_all = count_all.loc[sorted(count_all.index, key=nuc_key)]
    print("\nRows per nuclide / reaction  (long file):")
    print(count_all.to_string())
    print("\nRows per nuclide / reaction  (train file):")
    print(count_trn.reindex(count_all.index, fill_value=0).to_string())

    one_only = [n for n in count_all.index if (count_all.loc[n] == 0).any()]
    if one_only:
        print("\nNOTE nuclides that have only ONE reaction (fission or capture is missing):")
        print("  " + ", ".join(one_only))
        print("  (check whether a file is missing or was skipped, or whether that data really does not exist)")

    dups = {k: v for k, v in sources.items() if len(v) > 1}
    if dups:
        print("\nWARNING: several files map to the same nuclide + reaction (different libraries?):")
        for (n, r), fl in dups.items():
            print(f"  {n} {r}: {fl}")

    print_skipped(skipped)
    print("\nWritten:", OUT_LONG, "and", OUT_TRAIN)


def print_skipped(skipped):
    if skipped:
        print("\nSKIPPED FILES (rename them or fix them, then run again):")
        for n, why in skipped:
            print(f"  {n}: {why}")


if __name__ == "__main__":
    main()
```

```python
"""
train_and_test_model.py
"""
import warnings
from pathlib import Path

import joblib
import matplotlib
import numpy as np
import pandas as pd

matplotlib.use("Agg")
import matplotlib.pyplot as plt
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import r2_score
from sklearn.model_selection import GroupKFold

warnings.filterwarnings("ignore")

# ------------------------------ settings ------------------------------
SEED = 42
TEST_FRACTION = 0.20          # share of NUCLIDES kept for the final test
TEST_NUCLIDES = None          # or force them, e.g. ["Pu242", "Cm244"]
CV_FOLDS = 5                  # cross-validation folds (by nuclide)
FEATURES = ["Z", "N", "A", "fissility", "Z_even", "N_even", "reaction_code", "log10_E_MeV"]
TARGET = "log10_xs_barn"
ENERGY_BINS_MEV = [1e-3, 1e-2, 1e-1, 1.0, 20.0001]
ENERGY_LABELS = ["1-10 keV", "10-100 keV", "0.1-1 MeV", "1-20 MeV"]

try:
    FOLDER = Path(__file__).resolve().parent
except NameError:
    FOLDER = Path.cwd()


def banner(text):
    print("\n" + "=" * 70 + f"\n{text}\n" + "=" * 70)


# ============================ STEP 1: open the file ============================
banner("STEP 1  open train_data.csv")
df = pd.read_csv(FOLDER / "train_data.csv")
need = {"nuclide", "Z", "A", "N", "reaction", TARGET, "log10_E_MeV"}
missing = need - set(df.columns)
if missing:
    raise SystemExit(f"train_data.csv is missing columns: {missing}")
df = df.replace([np.inf, -np.inf], np.nan).dropna(subset=list(need)).reset_index(drop=True)
print(f"rows: {len(df)}   nuclides: {df.nuclide.nunique()}")
print(df.groupby(["nuclide", "reaction"]).size().unstack(fill_value=0).to_string())
if df.nuclide.nunique() < 6:
    raise SystemExit("Need at least 6 nuclides for a meaningful train/test split by nuclide.")

```

```python
# ============================ STEP 2: features ============================
banner("STEP 2  build the model inputs")
df["reaction_code"] = (df.reaction == "capture").astype(int)     # fission = 0, capture = 1
df["fissility"] = df.Z ** 2 / df.A                               # Z^2/A, classical fission parameter
df["Z_even"] = (df.Z % 2 == 0).astype(int)
df["N_even"] = (df.N % 2 == 0).astype(int)
print("inputs :", FEATURES)
print("target :", TARGET, "(log10 of the cross section in barn)")

```

```python
# ============================ STEP 3: split by nuclide ============================
banner("STEP 3  split into TRAIN and TEST nuclides")
nuclides = sorted(df.nuclide.unique())
if TEST_NUCLIDES:
    test_nuc = list(TEST_NUCLIDES)
else:
    rng = np.random.RandomState(SEED)
    n_test = max(2, int(round(TEST_FRACTION * len(nuclides))))
    test_nuc = sorted(str(x) for x in rng.choice(nuclides, size=n_test, replace=False))
train_nuc = [n for n in nuclides if n not in test_nuc]
train_df = df[df.nuclide.isin(train_nuc)].reset_index(drop=True)
test_df = df[df.nuclide.isin(test_nuc)].reset_index(drop=True)
print("TRAIN nuclides:", train_nuc)
print("TEST  nuclides:", test_nuc, " (never used for training or model choice)")
print(f"train rows: {len(train_df)}   test rows: {len(test_df)}")


# ============================ STEP 4: metrics ============================
def metrics(y_true_log, y_pred_log):
    y_true_log, y_pred_log = np.asarray(y_true_log, float), np.asarray(y_pred_log, float)
    err = y_pred_log - y_true_log
    yt, yp = 10 ** y_true_log, 10 ** y_pred_log
    rel = np.abs(10 ** err - 1.0)                                # |pred - true| / true
    r2 = r2_score(y_true_log, y_pred_log) if len(err) > 1 and np.var(y_true_log) > 0 else np.nan
    return {
        "n": len(err),
        "R2_log10": r2,
        "RMSE_log10": float(np.sqrt(np.mean(err ** 2))),
        "MAE_log10": float(np.mean(np.abs(err))),
        "bias_log10": float(np.mean(err)),
        "max_abs_err_log10": float(np.max(np.abs(err))),
        "MAPE_%": float(100 * np.mean(rel)),
        "median_APE_%": float(100 * np.median(rel)),
        "within_10%": float(np.mean(rel <= 0.10)),
        "within_20%": float(np.mean(rel <= 0.20)),
        "within_factor2": float(np.mean(np.abs(err) <= np.log10(2))),
        "RMSE_barn": float(np.sqrt(np.mean((yp - yt) ** 2))),
        "MAE_barn": float(np.mean(np.abs(yp - yt))),
    }


def metrics_by(frame, pred, by):
    rows = []
    for key, g in frame.assign(pred=pred).groupby(by, observed=True):
        m = metrics(g[TARGET], g["pred"])
        m[by if isinstance(by, str) else "group"] = key
        rows.append(m)
    out = pd.DataFrame(rows)
    first = by if isinstance(by, str) else "group"
    return out[[first] + [c for c in out.columns if c != first]]


def show(table):
    print(table.to_string(index=False, float_format=lambda v: f"{v:.4g}"))


# ---- simple baseline: average training curve per reaction (binned in log10 E)
class BinnedMeanBaseline:
    def __init__(self, n_bins=30):
        self.n_bins = n_bins

    def fit(self, d):
        self.edges = np.linspace(d.log10_E_MeV.min(), d.log10_E_MeV.max() + 1e-9, self.n_bins + 1)
        b = np.clip(np.digitize(d.log10_E_MeV, self.edges) - 1, 0, self.n_bins - 1)
        self.table = d.assign(b=b).groupby(["reaction_code", "b"])[TARGET].mean()
        self.global_mean = d[TARGET].mean()
        return self

    def predict(self, d):
        b = np.clip(np.digitize(d.log10_E_MeV, self.edges) - 1, 0, self.n_bins - 1)
        keys = list(zip(d.reaction_code, b))
        return np.array([self.table.get(k, self.global_mean) for k in keys])


# ============================ STEP 5: validate the ONE model ============================
banner("STEP 5  validate the model (cross-validation on TRAIN nuclides only)")
MODEL_NAME = "GradientBoostingRegression"


def make_model():
    # ONE model, fixed settings: gradient boosting regression (scikit-learn)
    return HistGradientBoostingRegressor(
        max_iter=500, learning_rate=0.05, max_leaf_nodes=20, min_samples_leaf=8,
        early_stopping=False, random_state=SEED)


n_splits = min(CV_FOLDS, train_df.nuclide.nunique())
gkf = GroupKFold(n_splits=n_splits)
cv_rows = []
for name in [MODEL_NAME, "Baseline_binned_mean"]:
    for fold, (tr, va) in enumerate(gkf.split(train_df, groups=train_df.nuclide)):
        tr_d, va_d = train_df.iloc[tr], train_df.iloc[va]
        if name == "Baseline_binned_mean":
            pred = BinnedMeanBaseline().fit(tr_d).predict(va_d)
        else:
            pred = make_model().fit(tr_d[FEATURES], tr_d[TARGET]).predict(va_d[FEATURES])
        m = metrics(va_d[TARGET], pred)
        m.update(model=name, fold=fold)
        cv_rows.append(m)
cv = pd.DataFrame(cv_rows)
cv_summary = (cv.groupby("model")[["R2_log10", "RMSE_log10", "MAE_log10", "MAPE_%",
                                   "median_APE_%", "within_20%", "within_factor2"]]
              .mean().sort_values("RMSE_log10").reset_index())
print(f"{n_splits}-fold cross-validation, each fold = nuclides the model has not seen (mean over folds):")
show(cv_summary)
best_name = MODEL_NAME
print(f"\nMODEL: {MODEL_NAME}  (no model selection; the baseline is only a yardstick)")

# ============================ STEP 6: train the chosen model ============================
banner("STEP 6  train the chosen model on all TRAIN nuclides")
model = make_model().fit(train_df[FEATURES], train_df[TARGET])
train_pred = model.predict(train_df[FEATURES])
print("training-set metrics (expected to look better than the test; do NOT report these as the result):")
show(pd.DataFrame([metrics(train_df[TARGET], train_pred)]))

# ============================ STEP 7: test ============================
banner("STEP 7  TEST on nuclides the model has never seen")
test_pred = model.predict(test_df[FEATURES])
base = BinnedMeanBaseline().fit(train_df)
base_pred = base.predict(test_df)
print(f"predicted {len(test_df)} points for {test_nuc}")

# ============================ STEP 8: test metrics ============================
banner("STEP 8  test metrics")
overall = pd.DataFrame([{"model": best_name, **metrics(test_df[TARGET], test_pred)},
                        {"model": "Baseline_binned_mean", **metrics(test_df[TARGET], base_pred)}])
print("OVERALL (model vs. baseline):")
show(overall)

by_reaction = metrics_by(test_df, test_pred, "reaction")
print("\nPER REACTION:")
show(by_reaction)

by_nuclide = metrics_by(test_df.assign(nr=test_df.nuclide + "_" + test_df.reaction), test_pred, "nr")
print("\nPER NUCLIDE and REACTION:")
show(by_nuclide)

test_df["energy_range"] = pd.cut(10 ** test_df.log10_E_MeV, ENERGY_BINS_MEV, labels=ENERGY_LABELS, right=False)
by_energy = metrics_by(test_df.dropna(subset=["energy_range"]), test_pred[test_df.energy_range.notna().to_numpy()],
                       "energy_range")
print("\nPER ENERGY RANGE:")
show(by_energy)

# ============================ STEP 9: save ============================
banner("STEP 9  save results")
out = FOLDER / "results"
out.mkdir(exist_ok=True)
joblib.dump({"model": model, "features": FEATURES, "model_name": best_name,
             "train_nuclides": train_nuc, "test_nuclides": test_nuc}, out / "model.joblib")
cv_summary.to_csv(out / "cv_summary.csv", index=False)
cv.to_csv(out / "cv_all_folds.csv", index=False)
overall.to_csv(out / "test_metrics_overall.csv", index=False)
by_reaction.to_csv(out / "test_metrics_by_reaction.csv", index=False)
by_nuclide.to_csv(out / "test_metrics_by_nuclide.csv", index=False)
by_energy.to_csv(out / "test_metrics_by_energy.csv", index=False)
pd.DataFrame({
    "nuclide": test_df.nuclide, "reaction": test_df.reaction,
    "energy_MeV": 10 ** test_df.log10_E_MeV,
    "xs_true_barn": 10 ** test_df[TARGET], "xs_pred_barn": 10 ** test_pred,
    "abs_pct_error": 100 * np.abs(10 ** (test_pred - test_df[TARGET]) - 1)}
).to_csv(out / "test_predictions.csv", index=False)

# plot 1: predicted vs true
fig, ax = plt.subplots(figsize=(5.5, 5.5))
for r, c in [("fission", "#1f4e79"), ("capture", "#c0392b")]:
    m = (test_df.reaction == r).to_numpy()
    ax.scatter(10 ** test_df[TARGET][m], 10 ** test_pred[m], s=8, alpha=.6, c=c, label=r)
lim = [10 ** test_df[TARGET].min() * .7, 10 ** test_df[TARGET].max() * 1.4]
ax.plot(lim, lim, "k--", lw=1)
ax.set_xscale("log"); ax.set_yscale("log"); ax.set_xlim(lim); ax.set_ylim(lim)
ax.set_xlabel("true cross section (barn)"); ax.set_ylabel("predicted (barn)")
ax.set_title("TEST nuclides: predicted vs true"); ax.legend()
plt.tight_layout(); plt.savefig(out / "plot_pred_vs_true.png", dpi=120); plt.close()

# plot 2: curves for the test nuclides
combos = test_df[["nuclide", "reaction"]].drop_duplicates().values.tolist()
cols = 2
rows_n = int(np.ceil(len(combos) / cols))
fig, axes = plt.subplots(rows_n, cols, figsize=(11, 3.6 * rows_n), squeeze=False)
for ax, (n, r) in zip(axes.ravel(), combos):
    m = ((test_df.nuclide == n) & (test_df.reaction == r)).to_numpy()
    e = 10 ** test_df.log10_E_MeV[m]
    ax.loglog(e, 10 ** test_df[TARGET][m], "-", c="#1f4e79", lw=1.3, label="true")
    ax.loglog(e, 10 ** test_pred[m], "--", c="#c0392b", lw=1.3, label="predicted")
    ax.set_title(f"{n} {r}"); ax.set_xlabel("energy (MeV)"); ax.set_ylabel("barn"); ax.legend()
for ax in axes.ravel()[len(combos):]:
    ax.axis("off")
plt.tight_layout(); plt.savefig(out / "plot_test_curves.png", dpi=110); plt.close()

print(f"saved in: {out}")
print("  model.joblib, cv_summary.csv, cv_all_folds.csv, test_metrics_*.csv, test_predictions.csv,")
print("  plot_pred_vs_true.png, plot_test_curves.png")
print("\nDONE.")
```
